```python
import asyncio
import nest_asyncio
import sys
import warnings
from playwright.async_api import async_playwright

# 1. Silence warnings and force the correct Windows Manager
warnings.filterwarnings("ignore", category=DeprecationWarning)
if sys.platform == 'win32':
    try:
        asyncio.set_event_loop_policy(asyncio.WindowsProactorEventLoopPolicy())
    except:
        pass # Already set

nest_asyncio.apply()

# 2. Define the launch with extra 'waiting' logic
async def start_session():
    print("Starting Playwright...")
    pw = await async_playwright().start()
    
    # We add a small timeout here to give Windows a chance to breath
    browser = await pw.chromium.launch(headless=False, args=["--no-sandbox"])
    
    page = await browser.new_page()
    print("✅ Browser is open!")
    return pw, browser, page

# 3. Execute
try:
    playwright_instance, browser_instance, page_instance = await start_session()
    await page_instance.goto("https://www.google.com")
    print(f"Currently viewing: {await page_instance.title()}")
except Exception as e:
    print(f"❌ Still failing. Error detail: {e}")
```
